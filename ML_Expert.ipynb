```python
from groq import Groq
from google.colab import userdata
key = userdata.get("GROG_API")

client= Groq(api_key=key)

print("Grog API connected successfully")
```

```python
models = client.models.list()
print("Available Groq Models:")
for model in models.data:
  print(model.id)
```

Test the Prompt

```python
response = client.chat.completions.create(
    model= "qwen/qwen3.8-27b",
    messages= [
        {"role": "system", "content": "Build an ML expert."},
        {"role": "user", "content": "Explain CNN (Convolutional Neural Network) in exactly 5 points, covering its introduction, architecture, working principle, real-world applications, and advantages and limitations. Use simple English, clear examples, and easy-to-understand explanations suitable for an engineering student."}
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response.choices[0])
print(response.choices[0].message.content)
```
