Import libraries

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score,confusion_matrix,classification_report, precision_score,recall_score
```

Load File

```python
df = pd.read_excel("/content/diabetesxl.xlsx")
df.head(7)
```

Null and Drop

```python
df.isnull().sum()
```

```python
df.dropna()
df.isnull().sum()
```

```python
df.dropna(inplace= True)
df.isnull().sum()
```

Duplicate and Drop

```python
df.duplicated().sum()
```

```python
print("Number of duplicate rows:",df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
print("Number of duplicate rows:",df.duplicated().sum())
```

Train and Test

```python
x= df[["age","mass", "insu" ,"plas"]]
y= df["class"]
```

```python
x= df[["age","mass","insu", "plas"]]
y = df["class"]
print("Feature shape:",x.shape, "Target shape:",y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size = 0.2, random_state = 42
)
print("Records for training:", len(x_train))
print("Records for testing:", len(x_test))
```

Train Random Forest

```python
Rand =  RandomForestClassifier()
```

```python
Rand= RandomForestClassifier(n_estimators=100, random_state=42)
Rand.fit(x_train, y_train)
y_predict= Rand.predict(x_test)
```

```python
y_predict = Rand.predict(x_test)
result= pd.DataFrame({
    "Actual": y_test,
    "Predicated" : y_predict
})
result.head(10)
```

```python
age = int(input("Enter Age: "))
mass= int(input("Enter mass: "))
insulin= int(input("Enter insulin: "))
plasma= int(input("Enter plasma level: "))
pred=Rand.predict([[age,mass,insulin,plasma]])

if pred == 'tested_positive':
  print("Please consult a Doctor")
else:
  print("Eat Kaju moreee")
```

```python
print("Accuracy:", accuracy_score(y_test,y_predict))
print("Precision:", precision_score(y_test,y_predict,pos_label="tested_positive"))
print("Recall:", recall_score(y_test,y_predict,pos_label="tested_positive"))
print("F1 Score:", f1_score(y_test,y_predict,pos_label="tested_positive"))
print("Confusion Matrix:\n", confusion_matrix(y_test,y_predict))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cm = confusion_matrix(y_test,y_predict)
sns.heatmap(cm,annot=True)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```
